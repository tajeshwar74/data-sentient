#Train_Test_Split


```python
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("/content/sample_data/500hits.csv", encoding="latin-1")
```

```python
df.head()
```

```python
X = df.drop(columns=['PLAYER', 'HOF'])
```

#Normalization & Standardization


```python
df = df.drop(columns=['PLAYER', 'CS'])
```

```python
X1 = df.iloc[:, 0:13]
```

```python
X2 = df.iloc[:, 0:13]
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaleStandard = StandardScaler()
```

```python
X1 = scaleStandard.fit_transform(X1)
```

```python
X1 = pd.DataFrame(X1, columns=['YRS', 'G', 'AB','R', 'H', '2B', '3B', 'HR', 'RBI', 'BB', 'SO', 'SB', 'BA'])
```

```python
X1.head()
```

```python
X1.describe().round(3)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaleMinMax = MinMaxScaler(feature_range=(0,1))
```

```python
X2 = scaleMinMax.fit_transform(X2)
```

```python
X2 = pd.DataFrame(X2, columns=['YRS', 'G', 'AB','R', 'H', '2B', '3B', 'HR', 'RBI', 'BB', 'SO', 'SB', 'BA'])
```

```python
X2.head().round(3)
```

```python
X2.describe().round(3)
```
